```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 200)

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("WEB_CHALLAN_DETAILS_filtered.csv")
```

```python
df = df[(df["TOTALAMOUNT"].notna()) & (df["TOTALAMOUNT"] != 0)]
```

```python
df.head()
```

```python
df["CREATEDDATE"] = pd.to_datetime(df["CREATEDDATE"])

daily_df = (
    df.groupby(pd.Grouper(key="CREATEDDATE", freq="D"))["TOTALAMOUNT"]
      .sum()
      .reset_index()
)
```

```python
daily_df = daily_df[(daily_df["TOTALAMOUNT"].notna()) & (daily_df["TOTALAMOUNT"] != 0)]
```

```python
count = (daily_df["TOTALAMOUNT"] == 0.0).sum()
print(count)
```

```python
daily_df["CREATEDDATE"] = pd.to_datetime(daily_df["CREATEDDATE"])
# train = monthly_amount.loc[monthly_amount['YearMonth'] >= '2022-04-01' and monthly_amount['YearMonth'] < '2025-06-01']
# test = monthly_amount.loc[monthly_amount['YearMonth'] >= '2025-06-01']
train = daily_df.loc[
    (daily_df["CREATEDDATE"] >= "2022-03-01") &
    (daily_df["CREATEDDATE"] < "2025-02-01")
]

# Test data
test = daily_df.loc[
    daily_df["CREATEDDATE"] >= "2025-02-01"
]
```

```python
def create_features_daily(df):
    df = df.copy()

    # date parts
    df["dayofweek"] = df["CREATEDDATE"].dt.dayofweek   # 0 = Monday
    df["day"] = df["CREATEDDATE"].dt.day
    df["month"] = df["CREATEDDATE"].dt.month
    df["year"] = df["CREATEDDATE"].dt.year

    # cyclical encoding for weekly seasonality (7 days)
    df["day_sin"] = np.sin(2 * np.pi * df["dayofweek"] / 7)
    df["day_cos"] = np.cos(2 * np.pi * df["dayofweek"] / 7)

    # optional: yearly seasonality (365 days)
    df["dayofyear"] = df["CREATEDDATE"].dt.dayofyear
    df["year_sin"] = np.sin(2 * np.pi * df["dayofyear"] / 365)
    df["year_cos"] = np.cos(2 * np.pi * df["dayofyear"] / 365)

    # lags (daily → smaller + medium + long)
    for l in [1, 2, 3, 7, 14, 30, 60, 90]:
        df[f"lag_{l}"] = df["TOTALAMOUNT"].shift(l)

    # rolling windows (adjusted for daily)
    for w in [7, 14, 30, 60, 90]:
        df[f"rolling_mean_{w}"] = df["TOTALAMOUNT"].shift(1).rolling(w).mean()

    # volatility
    df["rolling_std_7"] = df["TOTALAMOUNT"].shift(1).rolling(7).std()
    df["rolling_std_30"] = df["TOTALAMOUNT"].shift(1).rolling(30).std()

    # trend features
    df["diff_1"] = df["TOTALAMOUNT"].diff(1)
    df["diff_7"] = df["TOTALAMOUNT"].diff(7)
    df["pct_change_1"] = df["TOTALAMOUNT"].pct_change(1)

    return df
```

```python
train = create_features_daily(train)
test = create_features_daily(test)
```

```python
FEATURES = []

for col in train.columns:
    if col not in ["CREATEDDATE", "TOTALAMOUNT"]:
        # print(col)
        FEATURES.append(col)
        
TARGET = 'TOTALAMOUNT'

X_train = train[FEATURES]
y_train = train[TARGET]

X_test = test[FEATURES]
y_test = test[TARGET]
```

```python
import xgboost as xgb
from sklearn.model_selection import KFold, cross_val_score
import numpy as np

xgb_model = xgb.XGBRegressor(
    objective="reg:squarederror",
    max_depth=1,
    learning_rate=0.05,
    n_estimators=200,
    subsample=0.7,
    colsample_bytree=0.7,
    reg_alpha=5,
    reg_lambda=10,
    min_child_weight=5,
    random_state=42
)
y_train_log = np.log1p(y_train)

kf = KFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(
    xgb_model,
    X_train,
    y_train_log,
    cv=kf,
    scoring="neg_root_mean_squared_error"
)

print("RMSE per fold:", -scores)
print("Mean RMSE:", -scores.mean())
```

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import cross_val_score, KFold
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
import numpy as np

ridge_model = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("ridge", Ridge(alpha=10))
])

y_train_log = np.log1p(y_train)

kf = KFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(
    ridge_model,
    X_train,
    y_train_log,
    cv=kf,
    scoring="neg_root_mean_squared_error"
)
print("RMSE per fold:", -scores)
print("Mean RMSE:", -scores.mean())
```

```python
xgb_model.fit(X_train, y_train_log)
ridge_model.fit(X_train, y_train_log)
```

```python
import numpy as np

y_pred_xgb_log = xgb_model.predict(X_test)
y_pred_ridge_log = ridge_model.predict(X_test)

y_pred_ensemble_log = (y_pred_xgb_log + y_pred_ridge_log) / 2
```

```python
y_pred_ensemble = np.expm1(y_pred_ensemble_log)
```

```python
import pandas as pd
import numpy as np

# combine train + test predictions
all_dates = pd.concat([train["CREATEDDATE"], test["CREATEDDATE"]], ignore_index=True)
test_dates = test["CREATEDDATE"].reset_index(drop=True)
train_dates = train["CREATEDDATE"].reset_index(drop=True)

y_test = y_test.reset_index(drop=True)
actual = np.concatenate([y_train, y_test])
predicted = np.concatenate([np.full(len(y_train), np.nan), y_pred_ensemble])  # train has no preds

residuals = y_test - y_pred_ensemble
std = residuals.std()

upper = y_pred_ensemble + 2 * std
lower = y_pred_ensemble - 2 * std

#actual values of both train and test fields.

df_plot = pd.DataFrame({
    "date": all_dates,
    "actual": actual,
    "predicted": predicted,
})
monthly_df = (
    df_plot.set_index("date")
           .resample("M")
           .sum()
           .reset_index()
)

#predicted values shown on graph
test_plot = pd.DataFrame({
    "date": test_dates,
    "actual": y_test,
    "predicted": y_pred_ensemble
})

monthly_test = test_plot.set_index("date").resample("M").sum().reset_index()

#upper and lower bounds of prediction
interval_df = pd.DataFrame({
    "date": test_dates.reset_index(drop=True),
    "upper": upper,
    "lower": lower
})

monthly_interval = (
    interval_df.set_index("date")
               .resample("M")
               .sum()
               .reset_index()
)
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

plt.figure(figsize=(14,6))

plt.plot(
    monthly_df["date"],
    monthly_df["actual"],
    marker="o",
    label="Actual (Monthly)"
)

plt.plot(
    monthly_test["date"],
    monthly_test["predicted"],
    marker="s",
    linestyle="--",
    label="Predicted (Monthly)"
)

plt.plot(
    monthly_interval["date"],
    monthly_interval["upper"],
    color="green",
    linestyle=":",
    linewidth=1,
)

plt.plot(
    monthly_interval["date"],
    monthly_interval["lower"],
    color="green",
    linestyle=":",
    linewidth=1,
)

plt.fill_between(
    monthly_interval["date"],
    monthly_interval["lower"],
    monthly_interval["upper"],
    color="green",
    alpha=0.2,
    label="Prediction Range"
)

# Train/Test split
plt.axvline(
    train_dates.iloc[-1],
    color="black",
    linestyle=":",
    linewidth=2,
    label="Train/Test Split"
)

ax = plt.gca()

# show only YEAR on x-axis
ax.xaxis.set_major_locator(mdates.YearLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

ax.xaxis.set_minor_locator(mdates.MonthLocator())

plt.xticks(rotation=45)
plt.xlabel("Year")
plt.ylabel("Total Amount (Monthly Aggregated)")
plt.title("Monthly Actual vs Predicted (Year-wise view)")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
